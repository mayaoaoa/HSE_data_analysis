На своем наборе данных построить:
- `Violinplot` (скрипичный график) по 3 атрибутам
- `Histplot` с распределением и оттенком

### Теория

Я загрузила данные, затем выделила 3 метрических (`Grades`, `Sleep_Hours_Per_Day`, `Study_Hours_Per_Day`) и 2 категориальных (бинарный `Gender` и троичный `Stress_Level`) атрибута, которые будут использоваться для выполнения заданий. После чего построила скрипичные графики (`violinplot`) для визуализации распределения оценок в зависимости от категориальных переменных (уровня стресса и пола), гистограммы (`histplot`) с накоплением и без, и матрицы графиков (`pairplot`) для визуализации попарных взаимосвязей между 3-мя метрическими атрибутами и их окраски по категориальным переменным (по уровню стресса и по полу).

### Документация
Используемые библиотеки:
- pandas – основная библиотека для работы с табличными данными
- seaborn – основная библиотека для статистической визуализации, на которой построены все графики в работе
- matplotlib – библиотека для настройки фигур, заголовков, подписей осей и итогового отображения графиков

Методы:
- `sns.violinplot()` – строит скрипичный график, который сочетает в себе ящик с усами и оценку плотности распределения данных
- `sns.histplot()` – строит гистограмму для визуализации распределения одной или нескольких переменных
- `sns.pairplot()` – создает матрицу графиков рассеяния для всех попарных комбинаций числовых переменных, что позволяет быстро оценить взаимосвязи между ними
- `plt.figure()` – создает новую фигуру для графиков
- `plt.subplots()` – создает фигуру и набор подграфиков для более сложных компоновок
- `plt.title()`, `plt.xlabel()`, `plt.ylabel()` – задают заголовок графика и подписи осей
- `plt.suptitle()` – добавляет общий заголовок для всей фигуры
- `plt.grid()` – настройка сетки графика
- `plt.tight_layout()` – автоматически настраивает параметры компоновки, чтобы подписи и заголовки не накладывались друг на друга
- `plt.show()` – отображает все созданные графики

Параметры:
- `split` – разделяет "скрипку" для двух категорий
- `inner` – отображает квартили внутри графика
- `data`, `x`, `y`, `vars`, `hue` – выбор данных (основа) для построения графиков
- `bins` – количество столбцов
- `alpha` – прозрачность
- `kde` – добавляет оценку плотности распределения поверх гистограммы
- `multiple` – определяет способ отображения нескольких категорий: `stack` – с накоплением, `layer` – наложение слоев, `dodge` – группировка столбцов рядом
- `palette` – задает цветовую палитру


### Решение

#### Загрузка данных

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('student_lifestyle_dataset.csv')
```

#### **1:** Построить скрипичный график по 3 атрибутам

**Метрический:** `Grades`

**Бинарный:** `Gender` Male/Female

**Любой другой:** `Stress_Level` (Low/Moderate/High)

```python
plt.figure(figsize=(9, 5))

sns.violinplot(data=df, x='Stress_Level', y='Grades', hue='Gender', split=True, palette={'Male': 'lightskyblue', 'Female': 'orchid'}, inner='quart', fill=False)

plt.title('Распределение оценок по уровню стресса и полу', fontsize=14, fontweight='bold')
plt.xlabel('Stress_Level', fontsize=12, fontweight='bold')
plt.ylabel('Grades', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#### **2:** Построить `Histplot` с распределением и оттенком

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df, x='Grades', hue='Stress_Level', bins=15, alpha=0.6, kde = True, palette={'Low': 'limegreen', 'Moderate': 'orange', 'High': 'red'}, ax=axes[0])
axes[0].set_title('Распределение оценок по уровню стресса', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Grades', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Количество студентов', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.3)

sns.histplot(data=df, x='Grades', hue='Stress_Level', bins=15, alpha=0.6, kde = True, multiple='stack', palette={'Low': 'limegreen', 'Moderate': 'orange', 'High': 'red'}, ax=axes[1])  
axes[1].set_title('Распределение оценок по уровню стресса (с накоплением)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Grades', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Количество студентов', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#### **3:** Построить `Pairplot` (3 х 3 атрибутов)

**1. Метрические атрибуты:** `Study_Hours_Per_Day`, `Grades`, `Sleep_Hours_Per_Day`

**Разделение по:** `Stress_Level` (Low/Moderate/High)

```python
plt.figure(figsize=(9, 9))

sns.pairplot(data=df, vars=['Study_Hours_Per_Day', 'Grades', 'Sleep_Hours_Per_Day'], hue='Stress_Level', palette={'Low': 'limegreen', 'Moderate': 'orange', 'High': 'red'}, plot_kws={'alpha': 0.6, 's': 20})

plt.suptitle('Взаимосвязь количества учебных часов, оценок и количества сна (по уровню стресса)', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

**2. Метрические атрибуты:** `Study_Hours_Per_Day`, `Grades`, `Sleep_Hours_Per_Day`

**Разделение по:** `Gender` Male/Female

```python
plt.figure(figsize=(9, 9))

sns.pairplot(data=df, vars=['Study_Hours_Per_Day', 'Grades', 'Sleep_Hours_Per_Day'], hue='Gender', palette={'Male': 'lightskyblue', 'Female': 'orchid'}, plot_kws={'alpha': 1, 's': 15})

plt.suptitle('Взаимосвязь количества учебных часов, оценок и количества сна (в разрезе по полу)', fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Вывод

Были построены все требуемые графики.

**1.** Скрипичный график показал, что медианные оценки в разбивке по уровню стресса у мужчин и женщин практически совпадают, и при этом можно заметить, что чем выше уровень стресса у студента (любого пола), тем выше оценки.

**2.** По гистограмме без накопления можно четко заметить, что большинство студентов датасета имеют Высокий уровень стресса 😿. А график с накоплением лучше показывает распределение по каждой "отметке" оценки уровня стресса у студентов с этой оценкой.

**3.** По матрице графиков можно подробнее посмотреть эти взаимосвязи, а также взглянуть на 3й метрический параметр – количество часов сна в день. Из инетерсного, можно отметить, что студенты с Высоким уровнем стресса меньше всего спят и больше всего учатся. А распределения не по уровню стресса, а по полу студентов оказались примерно равными.
