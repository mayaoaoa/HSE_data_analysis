```python
import pandas as pd
import numpy as np
import scipy.stats as stats

df1 = pd.read_csv('samplesuperstore.csv', encoding='cp1252')
df2 = pd.read_csv('samplesuperstore.csv', encoding='cp1252')
df3 = pd.read_csv('samplesuperstore.csv', encoding='cp1252')
```

1. Рассчитать общие продажи с начала периода до текущей даты заказа

```python
df1['Order Date'] = pd.to_datetime(df1['Order Date'])
df1 = df1.sort_values('Order Date')

df1 = df1.groupby('Order Date')['Sales'].sum().reset_index()

df1['CumSales'] = df1['Sales'].cumsum()

pd.set_option('display.float_format', lambda x: f'{x:.2f}')
print(df1)
```

2. Рассчитать скользящее среднее продаж с окном в 3 месяца

```python
df2['Order Date'] = pd.to_datetime(df2['Order Date'])
df2 = df2.sort_values('Order Date')

df2['Order Month'] = df2['Order Date'].dt.to_period('M')
# print(df2['Order Month'])

df2 = df2.groupby('Order Month')['Sales'].sum().reset_index()
df2['Avg3M'] = df2['Sales'].rolling(window=3, min_periods=1).mean()

print(df2[['Order Month', 'Sales', 'Avg3M']])
```

3. Для каждой категории товаров и каждого года определить три товарные подкатегории с наибольшими общими продажами за год

```python
df3['Order Date'] = pd.to_datetime(df3['Order Date'])
df3 = df3.sort_values('Order Date')

df3['Order Year'] = df3['Order Date'].dt.year
# print(df3['Order Year'])

SumSub = df3.groupby(['Order Year', 'Category', 'Sub-Category'])['Sales'].sum().reset_index()
print(SumSub)

TopSub = (SumSub.sort_values(['Order Year', 'Category', 'Sales'], ascending=[True, True, False]).groupby(['Category', 'Order Year']).head(3).reset_index(drop=True))
print(TopSub)
```
