### Решение

#### Загрузка и предварительная обработка данных

```python
import pandas as pd

df = pd.read_csv('samplesuperstore.csv', sep=',', encoding='cp1252')
df = df.dropna()

print(df.shape)
```

#### 1. Создать новые атрибуты:
`year`, `month`, `quarter`, `day_of_week`, `day_of_year`, `is_month_end`, `season`: (`'Winter'`, `'Spring'`, `'Summer'`, `'Autumn'`)

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df = df.sort_values('Order Date')


df['year'] = df['Order Date'].dt.year
df['month_num'] = df['Order Date'].dt.month
df['month'] = df['Order Date'].dt.month_name()
df['quarter'] = df['Order Date'].dt.quarter
df['day_of_week_num'] = df['Order Date'].dt.dayofweek
df['day_of_week'] = df['Order Date'].dt.day_name()
df['day_of_year'] = df['Order Date'].dt.dayofyear
df['is_month_end'] = df['Order Date'].dt.is_month_end
df['season'] = df['month_num'].apply(lambda m: 'Winter' if m in [12, 1, 2] else
                                               'Spring' if m in [3, 4, 5] else
                                               'Summer' if m in [6, 7, 8] else
                                               'Autumn')

print("Созданные атрибуты (без воспомогательных):")
cols = ['Order Date', 'year', 'month', 'quarter', 'day_of_week', 'day_of_year', 'is_month_end', 'season']
display(df[cols].head(10))

print("\nis_month_end = True:")
display(df[df['is_month_end']][cols].head(5))
```

#### 2. Определить самый прибыльный день недели по средней прибыли и самый загруженный день по количеству заказов

```python
most_profit_day = df.groupby('day_of_week')['Profit'].mean()
most_order_day = df.groupby('day_of_week')['Order ID'].nunique()


print("most_profit_day:", most_profit_day.idxmax(), "– avg profit:", most_profit_day.max().round(2))
print("most_order_day:", most_order_day.idxmax(), "– orders:", most_order_day.max())
```

#### 3. Выделить данные только за выходные дни (суббота и воскресенье) и сравните средний чек с рабочими днями

```python
avg_weekend = df[df['day_of_week_num'] >= 5].groupby('Order ID')['Sales'].sum().mean().round(2)
avg_workday = df[df['day_of_week_num'] < 5].groupby('Order ID')['Sales'].sum().mean().round(2)

print("avg_weekend:", avg_weekend)
print("avg_workday:", avg_workday)
```

#### 4. Сгруппировать данные по кварталам и подсчитать для каждого квартала:
 - a. общий объем продаж
 - b. среднюю прибыльность
 - c. количество уникальных клиентов
 - d. определите лучший квартал по каждому из показателей

```python
task4a = df.groupby(['year', 'quarter'])['Sales'].sum().round(2)
task4b = df.groupby(['year', 'quarter'])['Profit'].mean().round(2)
task4c = df.groupby(['year', 'quarter'])['Customer ID'].nunique()


print("\na: общий объем продаж\n")
display(task4a)

print("\nb: средняя прибыльность")
display(task4b)

print("\nc: количество уникальных клиентов")
display(task4c)

print("\nd_a: лучший квартал по общему объему продаж:", task4a.idxmax()[0], "Q", task4a.idxmax()[1],
      "\nd_b: лучший квартал по средней прибыли:",  task4b.idxmax()[0], "Q", task4b.idxmax()[1],
      "\nd_c: лучший квартал по количеству уникальных клиентов:", task4c.idxmax()[0], "Q", task4c.idxmax()[1])
```
